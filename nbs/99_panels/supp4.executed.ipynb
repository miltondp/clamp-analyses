```r
library(here)
source(here("scripts/panels/style.R"))
set.seed(123)
cfg <- yaml::read_yaml(here("config.yaml"))
colours <- unlist(cfg$MODEL_COLORS)
names(colours)[names(colours)=="GenomicSuperSignature"] <- "GSSig"
paths <- c(pseudobulk=here("output/02_model_performance/00_pseudobulk/00_computational_timing/runtime_seed_totals.csv"),
 gtex=here("output/02_model_performance/01_gtex/00_computational_timing/runtime_per_fit.csv"))
pb <- fread(paths[["pseudobulk"]]); gt <- fread(paths[["gtex"]])
stopifnot(nrow(pb)==30L,all(pb$n_datasets==6L),all(pb$n_runs==6L),nrow(gt)==27L)
make_runtime <- function(d,value,label,log=FALSE) {
 d <- copy(d); d[, method := factor(method,levels=d[,.(med=median(get(value))),by=method][order(med),method])]
 p <- ggplot(d,aes(method,.data[[value]],fill=method)) +
 geom_boxplot(width=.55,outlier.shape=NA,linewidth=.25,alpha=.8) +
 geom_point(aes(shape=factor(seed)),position=position_jitter(width=.07,height=0,seed=123),size=1) +
 scale_shape_manual(values=c(21,22,24),guide="none") +
 scale_fill_manual(values=colours,guide="none") + labs(x=NULL,y=label) + nm_theme() +
 theme(axis.text.x=element_text(angle=45,hjust=1,size=MIN_FONT_PT),plot.margin=margin(2,2,2,2,"mm"))
 if(log) p <- p+scale_y_log10()
 p
}
a <- make_runtime(pb,"total_minutes","Six-cohort total time (min)",TRUE)
b <- make_runtime(gt,"elapsed_minutes","GTEx elapsed time (min)",TRUE)
c <- make_runtime(gt,"peak_rss_gb","GTEx peak memory (GiB)")
page <- plot_grid(nm_tag(a,"a"),nm_tag(b,"b"),nm_tag(c,"c"),nrow=1)
nm_sources(4,paths)
nm_export(list(page),4,"",height=73,heading=FALSE)
```
