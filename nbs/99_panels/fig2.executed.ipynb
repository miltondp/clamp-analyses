```r
long     <- fread_gm(panel_input[["benchmark_long"]])
long_box <- long[truth == "v0" & !is.na(cor)]

mean_order <- long_box[, .(m = mean(cor, na.rm = TRUE)), by = method]
setorder(mean_order, -m)
method_order <- as.character(mean_order$method)
long_box[, method := factor(method, levels = method_order)]

METHOD_COLORS <- MODEL_COLORS_RAW[method_order]
METHOD_COLORS[is.na(METHOD_COLORS)] <- "grey70"
names(METHOD_COLORS) <- method_order

mean_df <- long_box[, .(mean_cor = mean(cor, na.rm = TRUE),
                        max_cor  = max(cor, na.rm = TRUE)), by = method]
mean_df[, method := factor(method, levels = method_order)]

xpos <- setNames(seq_along(method_order), method_order)
other_methods <- setdiff(method_order, "CLAMPfull")

fmt_q <- function(q) {
  if (q >= 0.001) {
    sprintf('"%.3f"', q)
  } else {
    e_str    <- formatC(q, format = "e", digits = 1)
    parts    <- strsplit(e_str, "e")[[1]]
    mantissa <- trimws(parts[1])
    exp_val  <- as.integer(parts[2])
    sprintf('%s%%*%%10^{%d}', mantissa, exp_val)
  }
}

wide_ct_box <- dcast(long_box, dataset + cell_type ~ method, value.var = "cor")

comp_df_paired <- rbind(
  data.table(a = "CLAMPfull", b = other_methods[1:3]),
  data.table(a = "CLAMPbase", b = "PLIER")
)
comp_df_paired[, p_raw := mapply(function(a, b) {
  d <- wide_ct_box[!is.na(get(a)) & !is.na(get(b))]
  wilcox.test(d[[a]], d[[b]], paired = TRUE, alternative = "two.sided", exact = FALSE)$p.value
}, a, b)]
comp_df_paired[, q := p.adjust(p_raw, method = "BH")]

comp_df_paired[, p_one := mapply(function(a, b) {
  d <- wide_ct_box[!is.na(get(a)) & !is.na(get(b))]
  wilcox.test(d[[a]], d[[b]], paired = TRUE, alternative = "greater", exact = FALSE)$p.value
}, a, b)]
comp_df_paired[, q_one := p.adjust(p_one, method = "BH")]

assign_bracket_tiers <- function(comp_df, xpos) {
  comp_ord <- copy(comp_df)
  comp_ord[, x1 := xpos[a]]
  comp_ord[, x2 := xpos[b]]
  comp_ord[, left  := pmin(x1, x2)]
  comp_ord[, right := pmax(x1, x2)]
  comp_ord[, span  := right - left]
  setorder(comp_ord, span, q)

  levels_used <- list()
  comp_ord[, tier := 0L]

  for (i in seq_len(nrow(comp_ord))) {
    left  <- comp_ord$left[i]
    right <- comp_ord$right[i]
    tier  <- 1

    repeat {
      current <- if (tier <= length(levels_used)) levels_used[[tier]] else NULL
      overlaps <- FALSE

      if (!is.null(current)) {
        overlaps <- any(vapply(current, function(interval) {
          !(right < interval[1] || left > interval[2])
        }, logical(1)))
      }

      if (!overlaps) break
      tier <- tier + 1
    }

    comp_ord$tier[i] <- tier
    prior <- if (tier <= length(levels_used)) levels_used[[tier]] else list()
    levels_used[[tier]] <- c(prior, list(c(left, right)))
  }

  comp_ord
}

comp_ord <- assign_bracket_tiers(comp_df_paired, xpos)
n_tiers <- max(comp_ord$tier)
y_base  <- 1.08
y_step  <- 0.17
h       <- 0.02
y_max   <- y_base + (n_tiers - 1) * y_step + 0.14

plot_A <- ggplot(long_box, aes(method, cor, fill = method)) +
  geom_boxplot(width = 0.5, outlier.shape = NA, color = "black",
               linewidth = 0.25, alpha = 0.85) +
  geom_jitter(width = 0.10, size = 0.35, shape = 21, fill = "white",
              color = "#333333", stroke = 0.15, alpha = 0.75) +
  geom_point(data = mean_df, aes(x = method, y = mean_cor), shape = 23,
             size = 1.4, fill = "white", color = "black", stroke = 0.4,
             inherit.aes = FALSE) +
  geom_text(data = mean_df,
            aes(x = method, y = max_cor + 0.05, label = sprintf("%.3f", mean_cor)),
            size = 3.2, fontface = "bold", inherit.aes = FALSE) +
  scale_fill_manual(values = METHOD_COLORS, na.value = "grey70") +
  scale_y_continuous(breaks = seq(0, 1, 0.25), expand = expansion(mult = c(0.02, 0.02))) +
  coord_cartesian(ylim = c(0, y_max), clip = "on") +
  labs(x = NULL, y = "Max Pearson r") +
  theme_pub() +
  theme(legend.position = "none",
        axis.text.x = element_text(angle = 35, hjust = 1, size = AXIS_LABEL_SIZE),
        panel.grid.major = element_line(color = "grey88", linewidth = 0.25),
        panel.grid.minor = element_blank())

for (i in seq_len(nrow(comp_ord))) {
  row <- comp_ord[i, ]
  y   <- y_base + (row$tier - 1) * y_step

  if (
    (row$a == "CLAMPfull" & row$b == "CLAMPbase") |
    (row$a == "CLAMPbase" & row$b == "CLAMPfull")
  ) {
    y <- y + 0.02
  }

  lbl <- fmt_q(row$q)

  plot_A <- plot_A +
    annotate("segment", x = row$left,  xend = row$left,  y = y,       yend = y + h, linewidth = 0.25) +
    annotate("segment", x = row$left,  xend = row$right, y = y + h, yend = y + h, linewidth = 0.25) +
    annotate("segment", x = row$right, xend = row$right, y = y,       yend = y + h, linewidth = 0.25) +
    annotate("text",
             x = (row$left + row$right) / 2,
             y = y + h + 0.015,
             label = lbl,
             size = 3.0,
             vjust = 0,
             parse = TRUE,
             family = "sans")
}

options(repr.plot.width = 10, repr.plot.height = 3.2)
print(plot_A)
```

```r
FIG_W <- 180
FIG_H <- 170
stopifnot(FIG_W == 180, FIG_H == 170)

FS_TAG <- 8
FS_TITLE <- 6.5
FS_AXIS_TITLE <- 6.5
FS_AXIS_TEXT <- 5
FS_LEGEND <- 5
FS_LEGEND_TITLE <- 5.5
MIN_FONT_PT <- 5
MIN_GEOM_TEXT <- MIN_FONT_PT / .pt
DIVERGING_COLORS <- unlist(cfg$DIVERGING_COLORS)
DIVERGING_VALUES <- as.numeric(cfg$DIVERGING_VALUES)

theme_nm <- function(base_size = FS_AXIS_TEXT, base_line_size = 0.25) {
  theme_classic(base_size = base_size, base_family = "Helvetica") %+replace%
    theme(
      plot.title = element_text(size = FS_TITLE, face = "bold", hjust = 0.5,
                                margin = margin(b = 1.2)),
      axis.line = element_line(linewidth = base_line_size, colour = "black"),
      axis.ticks = element_line(linewidth = base_line_size, colour = "black"),
      axis.ticks.length = unit(1.1, "mm"),
      axis.text = element_text(size = FS_AXIS_TEXT, colour = "black"),
      axis.title = element_text(size = FS_AXIS_TITLE, face = "plain", colour = "black"),
      legend.text = element_text(size = FS_LEGEND),
      legend.title = element_text(size = FS_LEGEND_TITLE, face = "plain"),
      legend.key = element_blank(),
      legend.background = element_blank(),
      strip.background = element_blank(),
      strip.text = element_text(size = FS_TITLE, face = "bold", margin = margin(b = 0.7)),
      panel.grid = element_blank(),
      plot.margin = margin(1.3, 1.3, 1.3, 1.3)
    )
}

compact_ct <- function(x) {
  x <- abbreviate_ct(x)
  x <- gsub("Endothelial cells", "Endothelial", x, ignore.case = TRUE)
  x <- gsub("LymphaticEndothelial", "Lymphatic endothelial", x, fixed = TRUE)
  x <- gsub("Alveolar epithelium", "Alveolar epi.", x, ignore.case = TRUE)
  x <- gsub("Airway epithelium", "Airway epi.", x, ignore.case = TRUE)
  x <- gsub("Submucosal Gland", "Submucosal gland", x, ignore.case = TRUE)
  x <- gsub("Cardiomyocyte", "CM", x, ignore.case = TRUE)
  x
}

compact_axis_ct <- function(x) {
  x <- compact_ct(x)
  x <- gsub("Human ", "", x, fixed = TRUE)
  x <- gsub("Astrocyte", "Astro.", x, ignore.case = TRUE)
  x <- gsub("Microglial cell", "Microglia", x, ignore.case = TRUE)
  x <- gsub("Oligodendrocytes", "Oligodend.", x, ignore.case = TRUE)
  x <- gsub("Inhibitory neuron", "Inhibitory", x, ignore.case = TRUE)
  x <- gsub("Excitatory neuron", "Excitatory", x, ignore.case = TRUE)
  x <- gsub("Lymphatic endothelial", "Lymphatic EC", x, ignore.case = TRUE)
  x <- gsub("Endothelial cell", "Endothelial", x, ignore.case = TRUE)
  x <- gsub("Cardiomyocyte", "CM", x, ignore.case = TRUE)
  x <- gsub("Monocyte", "Mono.", x, ignore.case = TRUE)
  x <- gsub("Natural killer", "NK", x, ignore.case = TRUE)
  x <- gsub("Mesothelial cell", "Mesothelial", x, ignore.case = TRUE)
  trimws(x)
}

compact_lv_label <- function(x) {
  vapply(strsplit(x, " - ", fixed = TRUE), function(parts) {
    lv <- tail(parts, 1)
    lv <- if (grepl("GM[0-9]+", lv)) sub(".*(GM[0-9]+).*", "\\1", lv) else lv
    ct <- compact_axis_ct(parts[1])
    if (length(parts) == 1 || identical(ct, lv)) lv else paste(ct, lv, sep = "\n")
  }, character(1), USE.NAMES = FALSE)
}

compact_lv_label_vertical <- function(x) {
  vapply(strsplit(x, " - ", fixed = TRUE), function(parts) {
    lv <- tail(parts, 1)
    lv <- if (grepl("GM[0-9]+", lv)) sub(".*(GM[0-9]+).*", "\\1", lv) else lv
    gene_set <- if (length(parts) >= 3) parts[length(parts) - 1] else parts[1]
    gene_set <- compact_axis_ct(gene_set)
    gene_set <- gsub("^Astro.*L1.*$", "Astro. L1", gene_set)
    gene_set <- gsub("^Micro.*L1-6.*$", "Microglia L1-6", gene_set)
    gene_set <- gsub("Central memory CD4\\+ T cell", "Central mem. CD4", gene_set)
    gene_set <- gsub("Memory CD8\\+ T cell", "Memory CD8", gene_set)
    gene_set <- gsub("Type 3 dendritic cell \\(DC3\\)", "DC3", gene_set)
    gene_set <- gsub("Cytotoxic CD4\\+ T cell", "Cytotoxic CD4", gene_set)
    gene_set <- gsub("Natural killer T ?\\(NKT\\) cell", "NKT cell", gene_set)
    gene_set <- gsub("Natural killer cell", "NK cell", gene_set)
    gene_set <- gsub("GABAergic neuron", "GABAergic", gene_set)
    gene_set <- gsub("Cardiac progenitor cell", "Cardiac prog.", gene_set)
    gene_set <- gsub("Ciliated epithelial cell", "Ciliated epi.", gene_set)
    gene_set <- gsub("Gonadal endothelial", "Gonadal EC", gene_set)
    gene_set <- gsub("Pan-endothelial", "Pan-EC", gene_set)
    gene_set <- gsub("Type II pneumocyte", "Type II pneumo.", gene_set)
    gene_set <- gsub("Leydig precursor cell", "Leydig prec.", gene_set)
    if (identical(gene_set, lv)) lv else paste(gene_set, lv)
  }, character(1), USE.NAMES = FALSE)
}

lv_only_label <- function(x) {
  ifelse(grepl("GM[0-9]+", x), sub(".*(GM[0-9]+).*", "\\1", x),
         compact_lv_label(x))
}

compact_hard_title <- function(x) {
  x <- gsub("_", " ", x, fixed = TRUE)
  x <- gsub("B cells vs\\. T cells", "B vs. T cells", x)
  x <- gsub("CD4\\+ T cells vs\\. CD8\\+ T\n?cells", "CD4+ vs. CD8+ T", x)
  x <- gsub("Oligodendrocyte\nPrecursor Cells vs\\.\nOligodendrocytes",
            "OPC vs. oligodendrocytes", x)
  x <- gsub("Astrocytes vs\\. Microglia", "Astrocytes vs.\nmicroglia", x)
  x <- gsub("Inhibitory neurons vs\\.\nExcitatory neurons",
            "Inhibitory vs. excitatory", x)
  x <- gsub("AtrialCM vs\\.\nVentricularCM", "Atrial vs.\nventricular CM", x)
  x
}

compact_region <- function(x) {
  x <- gsub("Cervical spinal cord", "Cervical cord", x, ignore.case = TRUE)
  x <- gsub("Gastroesophageal junction", "GE junction", x, ignore.case = TRUE)
  x <- gsub("Substantia nigra", "Substantia nigra", x, ignore.case = TRUE)
  x <- gsub("Nucleus accumbens", "N. accumbens", x, ignore.case = TRUE)
  x <- gsub("Cerebellar hemisphere", "Cerebellar hemi.", x, ignore.case = TRUE)
  x <- gsub("Subcutaneous", "Subcut.", x, ignore.case = TRUE)
  x <- gsub("Visceral", "Visc.", x, ignore.case = TRUE)
  x <- gsub("Coronary", "Coron.", x, ignore.case = TRUE)
  x <- gsub("Transverse", "Transv.", x, ignore.case = TRUE)
  x <- gsub("Muscularis", "Muscul.", x, ignore.case = TRUE)
  x <- gsub("Left ventricle", "Left ventr.", x, ignore.case = TRUE)
  x <- gsub("Atrial appendage", "Atrial app.", x, ignore.case = TRUE)
  x
}

add_tag <- function(p, label, x = 0.002, y = 0.998) {
  ggdraw() +
    draw_plot(p, x = 0, y = 0, width = 1, height = 1) +
    draw_label(label, x = x, y = y, hjust = 0, vjust = 1,
               size = FS_TAG, fontface = "bold", fontfamily = "Helvetica")
}

# Four biologically difficult pairs retained in the main figure.
PAIR_GROUPS <- c(1, 2, 4, 6)
PAIR_LEGEND_ORDER <- c(
  "B cell", "T cell", "CD4+ T cell", "CD8+ T cell",
  "Astrocyte", "Microglia", "Atrial CM", "Ventricular CM",
  "Shared marker"
)
PAIR_SET_COLORS <- c(
  "B cell" = "#0072B2",
  "T cell" = "#D55E00",
  "CD4+ T cell" = "#009E73",
  "CD8+ T cell" = "#CC79A7",
  "Astrocyte" = "#56B4E9",
  "Microglia" = "#E69F00",
  "Atrial CM" = "#6A3D9A",
  "Ventricular CM" = "#B15928",
  "Shared marker" = "#555555"
)
PAIR_DISPLAY_TITLES <- c(
  "1" = "B vs. T cells",
  "2" = "CD4+ vs. CD8+ T cells",
  "4" = "Astrocyte vs. microglia",
  "6" = "Atrial vs. ventricular CM"
)
PAIR_DISPLAY_TITLES_SMALL <- c(
  "1" = "B vs.\nT cells",
  "2" = "CD4+ vs.\nCD8+ T cells",
  "4" = "Astrocyte vs.\nmicroglia",
  "6" = "Atrial vs.\nventricular CM"
)

canonical_pair_set <- function(x) {
  x <- gsub("_", " ", as.character(x), fixed = TRUE)
  x <- trimws(x)
  x <- sub("^B( cells?)?$", "B cell", x)
  x <- sub("^T( cells?)?$", "T cell", x)
  x <- sub("^CD4\\+?( T)?( cells?)?$", "CD4+ T cell", x)
  x <- sub("^CD8\\+?( T)?( cells?)?$", "CD8+ T cell", x)
  x <- sub("^(Ast|Astro\\.?s?|Astrocytes?)$", "Astrocyte", x)
  x <- sub("^(Mic|Microglia)$", "Microglia", x)
  x <- sub("^Atrial ?CM$", "Atrial CM", x)
  x <- sub("^Ventricular ?CM$", "Ventricular CM", x)
  x
}
related_corr_pub <- fread_gm(panel_input[["related_corr"]])

pair_axis_label <- function(x) {
  x <- compact_axis_ct(x)
  x <- sub(" cells$", "", x)
  x <- sub("^VentricularCM$", "Ventric. CM", x)
  x <- sub("^AtrialCM$", "Atrial CM", x)
  x
}

pair_y_axis_label <- function(x) {
  x <- pair_axis_label(x)
  x[x == "B"] <- "B cell"
  x[x == "T"] <- "T cell"
  x
}

make_pair_pub <- function(gid) {
  d <- copy(related_corr_pub[group_id == gid])
  members <- unique(d$assigned_cell_type)
  member_lv <- d$LV[match(members, d$assigned_cell_type)]
  d <- d[cell_type %in% members]
  axis_lab <- pair_axis_label(members)
  row_levels <- pair_y_axis_label(members)
  col_levels <- paste0(axis_lab, "\n", member_lv)
  d[, row_label := factor(pair_y_axis_label(cell_type), levels = row_levels)]
  d[, col_label := factor(paste0(pair_axis_label(assigned_cell_type), "\n", LV),
                          levels = col_levels)]
  title_lab <- PAIR_DISPLAY_TITLES_SMALL[[as.character(gid)]]

  ggplot(d, aes(col_label, row_label, fill = cor)) +
    geom_tile(colour = "white", linewidth = 0.25) +
    scale_fill_gradientn(colours = DIVERGING_COLORS, values = DIVERGING_VALUES,
                         limits = c(-1, 1), name = "Pearson r") +
    scale_colour_manual(values = setNames(c("white", "black"), c("TRUE", "FALSE")),
                        guide = "none") +
    scale_x_discrete(expand = c(0, 0)) +
    scale_y_discrete(expand = c(0, 0)) +
    coord_fixed() +
    labs(x = NULL, y = NULL, title = title_lab) +
    theme_nm() +
    theme(axis.line = element_blank(), axis.ticks = element_blank(),
          axis.text.x = element_text(size = MIN_FONT_PT, lineheight = 0.9),
          axis.text.y = element_text(size = MIN_FONT_PT, margin = margin(r = 0.5)),
          plot.title = element_text(size = 5.4, face = "bold", hjust = 0.5,
                                    lineheight = 0.9, margin = margin(b = 0.5)),
          plot.margin = margin(0.5, 0.6, 0.5, 0.6))
}

pair_pub_panels <- lapply(PAIR_GROUPS, make_pair_pub)
pair_corr_legend <- get_legend(
  pair_pub_panels[[1]] +
    guides(fill = guide_colourbar(title.position = "top", title.hjust = 0.5,
                                  barwidth = unit(13, "mm"),
                                  barheight = unit(1.4, "mm"))) +
    theme(legend.position = "bottom", legend.direction = "horizontal",
          legend.title = element_text(size = 5),
          legend.text = element_text(size = MIN_FONT_PT),
          legend.margin = margin(0, 0, 0, 0))
)
pair_pub_body <- wrap_plots(lapply(pair_pub_panels, function(p)
  p + theme(legend.position = "none")), ncol = 2)
plot_D_pair_pub <- ggdraw() +
  draw_plot(pair_pub_body, x = 0, y = 0.24, width = 1, height = 0.76) +
  draw_grob(pair_corr_legend, x = 0.18, y = 0.19,
            width = 0.64, height = 0.10)

comp_ord_pub <- copy(comp_ord[!(a == "CLAMPfull" & b == "CoGAPS")])
comp_ord_pub <- assign_bracket_tiers(comp_ord_pub, xpos)
y_base_pub <- 1.15
y_step_pub <- 0.115
h_pub <- 0.014
y_max_pub <- y_base_pub + (max(comp_ord_pub$tier) - 1) * y_step_pub + 0.18

plot_A_pub <- ggplot(long_box, aes(method, cor, fill = method)) +
  geom_boxplot(width = 0.52, outlier.shape = NA, color = "black",
               linewidth = 0.22, alpha = 0.85) +
  geom_jitter(width = 0.2, size = 0.24, shape = 21, fill = "white",
              color = "#333333", stroke = 0.1, alpha = 0.65) +
  geom_point(data = mean_df, aes(x = method, y = mean_cor), shape = 23,
             size = 1.05, fill = "white", color = "black", stroke = 0.28,
             inherit.aes = FALSE) +
  geom_text(data = mean_df,
            aes(x = method, y = max_cor + 0.035, label = sprintf("%.3f", mean_cor)),
            size = MIN_GEOM_TEXT, fontface = "bold", family = "Helvetica",
            inherit.aes = FALSE) +
  scale_fill_manual(values = METHOD_COLORS, na.value = "grey70") +
  scale_y_continuous(breaks = c(0,.5,1),
                     labels = c("0", "0.5", "1"),
                     expand = expansion(mult = c(0.01, 0.01))) +
  coord_cartesian(ylim = c(0, y_max_pub), clip = "off") +
  labs(x = NULL, y = "Max Pearson r") +
  theme_nm() +
  theme(legend.position = "none",
        axis.text.x = element_text(angle = 45, hjust = 1, vjust = 1, size = MIN_FONT_PT),
        axis.title.y = element_text(size = 5.7),
        plot.margin = margin(4.5, 0.8, 0.6, 2.5))

for (i in seq_len(nrow(comp_ord_pub))) {
  stat_row <- comp_ord_pub[i]
  y <- y_base_pub + (stat_row$tier - 1) * y_step_pub
  if (setequal(c(stat_row$a, stat_row$b), c("CLAMPfull", "CLAMPbase"))) y <- y + 0.012
  plot_A_pub <- plot_A_pub +
    annotate("segment", x = stat_row$left, xend = stat_row$left,
             y = y, yend = y + h_pub, linewidth = 0.22) +
    annotate("segment", x = stat_row$left, xend = stat_row$right,
             y = y + h_pub, yend = y + h_pub, linewidth = 0.22) +
    annotate("segment", x = stat_row$right, xend = stat_row$right,
             y = y, yend = y + h_pub, linewidth = 0.22) +
    annotate("text", x = (stat_row$left + stat_row$right) / 2,
             y = y + h_pub + 0.01, label = fmt_q(stat_row$q),
             size = MIN_GEOM_TEXT, vjust = 0, parse = TRUE, family = "Helvetica")
}

plot_statistics[, facet_label := factor(sprintf("%s\nr=%.2f  MAE=%.3f",
  c("1k1k","Perez","Heart","Xiong","Mathys","Lung")[match(as.character(panel_label), c("PBMC 1k1k","PBMC Perez","Heart Datar","Brain Xiong","Brain Mathys","Lung Sikkema"))], mean_pearson_r, mean_mae))]
facet_lookup <- setNames(as.character(plot_statistics$facet_label), as.character(plot_statistics$panel_label))
scatter_data[, facet_label := factor(facet_lookup[as.character(panel_label)],
  levels=as.character(plot_statistics$facet_label))]
plot_C_pub <- ggplot(scatter_data, aes(x = observed, y = predicted)) +
  geom_abline(slope = 1, intercept = 0, color = "grey25",
              linetype = "dashed", linewidth = 0.25) +
  ggrastr::rasterise(geom_point(aes(color = panel_label), alpha = 0.28, size = 0.16,
             show.legend = FALSE), dpi=300) +
  facet_wrap(~facet_label, ncol = 3) +
  coord_cartesian(xlim = c(0, 1), ylim = c(0, 1), expand = FALSE) +
  scale_x_continuous(breaks = c(0,.5,1),
                     labels = c("0", "0.5", "1")) +
  scale_y_continuous(breaks = c(0,.5,1),
                     labels = c("0", "0.5", "1")) +
  scale_color_manual(values = DATASET_COLORS) +
  labs(x = "True cell proportion", y = "Predicted cell proportion") +
  theme_nm() +
  theme(aspect.ratio = 0.48,
        axis.text = element_text(size = MIN_FONT_PT),
        axis.title = element_text(size = 6),
        panel.spacing.x = unit(1.0, "mm"),
        panel.spacing.y = unit(1.7, "mm"),
        strip.text = element_text(size = 5, face = "bold", lineheight=.85, margin = margin(b = 0.3)),
        plot.margin = margin(2.0, 0.3, 0.5, 0.7))

build_module_pub <- function(d) {
  ds <- d$dataset[1]
  ds_label <- if (ds %in% names(DATASET_LABELS)) DATASET_LABELS[ds] else ds
  d[, marker_row_label := compact_axis_ct(marker_row_label)]
  d[, lv_col_label := compact_lv_label_vertical(lv_col_label)]
  row_levels <- unique(d[order(marker_row_rank), marker_row_label])
  col_levels <- unique(d[order(lv_col_rank), lv_col_label])
  d[, marker_row_label := factor(marker_row_label, levels = row_levels)]
  d[, lv_col_label := factor(lv_col_label, levels = col_levels)]
  diagonal <- d[diag_recovered == TRUE]
  z_lim <- d$Z_LIM[1]
  q_lim <- d$Q_LIM[1]
  n_samples <- d$n_samples[1]

  p <- ggplot(d, aes(lv_col_label, marker_row_label)) +
    geom_point(aes(size = neg_log10_fdr, color = row_effect)) +
    {if (nrow(diagonal)) geom_point(data = diagonal, aes(size = neg_log10_fdr),
                                    shape = 1, color = "black", stroke = 0.55)} +
    scale_size_continuous(limits = c(0, q_lim), range = c(0.25, 2.15),
                          name = "Combined ORA -log10(FDR)") +
    scale_color_gradientn(colors = GRID_COLORS, values = GRID_VALUES,
                          limits = c(-z_lim, z_lim), name = "GM effect") +
    scale_x_discrete(drop = FALSE) +
    scale_y_discrete(drop = FALSE) +
    coord_fixed() +
    labs(x = NULL, y = NULL, title = paste0(ds_label, " (n = ", n_samples, ")")) +
    theme_nm() +
    theme(panel.grid.major = element_line(color = "grey90", linewidth = 0.18),
          axis.line = element_blank(), axis.ticks = element_blank(),
          axis.text.x = element_text(angle = 45, hjust = 1, vjust = 1,
                                     size = MIN_FONT_PT, lineheight = 0.88),
          axis.text.y = element_text(size = MIN_FONT_PT),
          plot.title = element_text(size = 5.7, face = "bold", hjust = 0.5,
                                    margin = margin(b = 0.5)),
          plot.margin = margin(0.5, 0.7, 0.5, 0.7))
  list(plot = p, n_cat = length(row_levels))
}

module_pub_blocks <- lapply(split(copy(module_ready), module_ready$dataset), build_module_pub)
module_pub_order <- order(-vapply(module_pub_blocks, `[[`, "n_cat", FUN.VALUE = numeric(1)))
module_pub_blocks <- module_pub_blocks[module_pub_order]
module_pub_panels <- lapply(module_pub_blocks, `[[`, "plot")
module_pub_widths <- sqrt(vapply(module_pub_blocks, `[[`, "n_cat", FUN.VALUE = numeric(1)))

top_idx <- 1:3
bottom_idx <- 4:6
top_sum <- sum(module_pub_widths[top_idx])
bottom_sum <- sum(module_pub_widths[bottom_idx])
legend_pad <- max(0.8, top_sum - bottom_sum)

module_legend <- get_legend(
  module_pub_panels[[1]] +
    guides(color = guide_colorbar(order = 1, title.position = "top",
                                  barwidth = unit(16, "mm"),
                                  barheight = unit(1.5, "mm")),
           size = guide_legend(order = 2, title.position = "top", nrow = 1)) +
    theme(legend.position = "bottom", legend.direction = "horizontal",
          legend.box = "vertical", legend.box.just = "center",
          legend.spacing.y = unit(0.5, "mm"),
          legend.box.margin = margin(0, 0, 0, 0),
          legend.key.width = unit(2.1, "mm"), legend.key.height = unit(1.8, "mm"),
          legend.text = element_text(size = MIN_FONT_PT),
          legend.title = element_text(size = 5))
)

C_top <- plot_grid(plotlist = lapply(module_pub_panels[top_idx],
                                     function(p) p + theme(legend.position = "none")),
                   nrow = 1, rel_widths = module_pub_widths[top_idx])
C_bottom <- plot_grid(
  plotlist = c(lapply(module_pub_panels[bottom_idx],
                      function(p) p + theme(legend.position = "none")),
               list(ggdraw() + draw_grob(module_legend, x = 0.08, y = 0.02,
                                         width = 0.84, height = 0.96))),
  nrow = 1, rel_widths = c(module_pub_widths[bottom_idx], legend_pad)
)
C_main <- plot_grid(C_top, C_bottom, ncol = 1,
                    rel_heights = c(max(module_pub_widths[top_idx]),
                                    max(module_pub_widths[bottom_idx])))

HARD_EFFECT_LIM <- ceiling(
  max(abs(hard_ready[group_id %in% PAIR_GROUPS, row_effect])) * 10
) / 10

build_hard_pub <- function(d) {
  d[, marker_row_label := compact_axis_ct(marker_row_label)]
  d[, lv_col_label := gsub("\n", " ", compact_lv_label(lv_col_label),
                            fixed = TRUE)]
  d[, lv_col_label := sub(".*(GM[0-9]+).*", "\\1", lv_col_label)]
  row_levels <- unique(d[order(marker_row_rank), marker_row_label])
  col_levels <- unique(d[order(lv_col_rank), lv_col_label])
  d[, marker_row_label := factor(marker_row_label, levels = row_levels)]
  d[, lv_col_label := factor(lv_col_label, levels = col_levels)]
  z_lim <- HARD_EFFECT_LIM
  q_lim <- d$Q_LIM[1]
  ggplot(d, aes(lv_col_label, marker_row_label)) +
    geom_point(aes(size = neg_log10_fdr, fill = row_effect),
               shape = 21, colour = "grey30", stroke = 0.35) +
    scale_size_continuous(limits = c(0, q_lim), range = c(0.35, 1.9),
                          name = expression("Azimuth ORA " * -log[10](FDR)),
                          breaks = c(20, 40, 60)) +
    scale_fill_gradientn(colors = GRID_COLORS, values = GRID_VALUES,
                         limits = c(-z_lim, z_lim),
                         breaks = seq(-z_lim, z_lim, length.out = 5),
                         name = "GM effect") +
    coord_fixed() +
    labs(x = NULL, y = NULL,
         title = PAIR_DISPLAY_TITLES_SMALL[[as.character(d$group_id[1])]]) +
    theme_nm() +
    theme(panel.grid.major = element_line(color = "grey90", linewidth = 0.18),
          axis.line = element_blank(), axis.ticks = element_blank(),
          axis.text.x = element_text(angle = 45, hjust = 1, size = MIN_FONT_PT,
                                     lineheight = 0.88),
          axis.text.y = element_text(size = MIN_FONT_PT),
          plot.title = element_text(size = 5.1, face = "bold", hjust = 0.5,
                                    lineheight = 0.9, margin = margin(b = 0.4)),
          plot.margin = margin(0.4, 0.5, 0.4, 0.5))
}

hard_pub_selected <- copy(hard_ready[group_id %in% PAIR_GROUPS])
hard_pub_panels <- lapply(PAIR_GROUPS, function(gid)
  build_hard_pub(copy(hard_pub_selected[group_id == gid])))
plot_D_pub <- C_main
hard_pub_legend <- get_legend(
  hard_pub_panels[[1]] +
    guides(fill = guide_colourbar(order = 1, title.position = "top",
                                  title.hjust = 0.5,
                                  barwidth = unit(13, "mm"),
                                  barheight = unit(1.4, "mm")),
           size = guide_legend(order = 2, title.position = "top",
                               title.hjust = 0.5, nrow = 1,
                               override.aes = list(fill = "grey55",
                                                   colour = "grey20"))) +
    theme(legend.position = "bottom", legend.box = "vertical",
          legend.box.just = "center", legend.spacing.y = unit(0.2, "mm"),
          legend.title = element_text(size = 5),
          legend.text = element_text(size = MIN_FONT_PT),
          legend.key.width = unit(2.8, "mm"),
          legend.key.height = unit(1.8, "mm"),
          legend.margin = margin(0, 0, 0, 0))
)
hard_pub_body <- wrap_plots(lapply(hard_pub_panels, function(p)
  p + theme(legend.position = "none")), ncol = 2)
plot_E_pub <- ggdraw() +
  draw_plot(hard_pub_body, x = 0, y = 0.24, width = 1, height = 0.76) +
  draw_grob(hard_pub_legend, x = 0.12, y = 0.13,
            width = 0.76, height = 0.16)

build_confusion_pub <- function(tissue_name, compact = FALSE) {
  d <- copy(subtissue_confusion[Tissue == tissue_name])
  order_dt <- unique(d[, .(True_SMTSD, True_Display)])
  lvls <- order_dt$True_SMTSD
  display <- setNames(compact_region(strip_parens(order_dt$True_Display)),
                      order_dt$True_SMTSD)
  d[, True_SMTSD := factor(True_SMTSD, levels = rev(lvls))]
  d[, Predicted_SMTSD := factor(Predicted_SMTSD, levels = lvls)]
  n_classes <- length(lvls)
  d[, show_label := (n_classes <= 3) | (True_SMTSD == Predicted_SMTSD) |
                      (Row_Proportion >= 0.10)]
  title <- strip_parens(subtissue_anatomical[Tissue == tissue_name]$Tissue_Display)

  ggplot(d, aes(Predicted_SMTSD, True_SMTSD, fill = Row_Proportion)) +
    geom_tile(color = "white", linewidth = 0.25) +
    scale_fill_gradientn(colours = CONFUSION_COLORS, limits = c(0, 1),
                         name = "True\nfraction") +
    scale_color_manual(values = c(`TRUE` = "white", `FALSE` = "black")) +
    scale_x_discrete(labels = display, expand = expansion(add = 0)) +
    scale_y_discrete(labels = display, expand = expansion(add = 0)) +
    coord_fixed() +
    labs(x = NULL, y = NULL, title = title) +
    theme_nm() +
    theme(panel.border = element_rect(fill = NA, color = "grey25", linewidth = 0.22),
          axis.line = element_blank(), axis.ticks = element_blank(),
          axis.text.x = element_text(angle = 45, hjust = 1,
                                     size = MIN_FONT_PT,
                                     margin = margin(t = if (compact) -1 else 0)),
          axis.text.y = element_text(size = MIN_FONT_PT,
                                     margin = margin(r = if (compact) -1 else 0)),
          plot.title = element_text(size = if (compact) 5.1 else 5.8,
                                    face = "plain", hjust = 0.5,
                                    margin = margin(b = 0.5)),
          plot.margin = if (compact) margin(0.2, 0.2, 0.2, 0.2)
                        else margin(0.5, 0.5, 0.5, 0.5))
}

brain_pub <- build_confusion_pub("Brain") +
  labs(x = "Predicted region", y = "True region") +
  theme(axis.title = element_text(size = 5.8))
small_pub <- lapply(SMALL_TISSUE_ORDER, build_confusion_pub, compact = TRUE)
confusion_legend_pub <- get_legend(
  brain_pub +
    guides(fill = guide_colorbar(title.position = "top",
                                 barheight = unit(14, "mm"),
                                 barwidth = unit(1.8, "mm"))) +
    theme(legend.position = "right", legend.text = element_text(size = MIN_FONT_PT),
          legend.title = element_text(size = 5, hjust = 0.5),
          legend.key.height = unit(2.3, "mm"))
)
small_grid_pub <- wrap_plots(lapply(small_pub, function(p) p + theme(legend.position = "none")),
                             ncol = 2)
small_x_pub <- ggdraw() + draw_label("Predicted region", size = 5.8,
                                     fontfamily = "Helvetica")
small_block_inner_pub <- plot_grid(small_grid_pub, small_x_pub,
                             ncol = 1, rel_heights = c(1, 0.07))
small_block_pub <- ggdraw() +
  draw_plot(small_block_inner_pub, x = 0.052, y = 0, width = 0.948, height = 1) +
  draw_label("True region", x = 0.17, y = 0.53, size = 5.8, angle = 90,
             fontfamily = "Helvetica")
plot_F_pub <- plot_grid(brain_pub + theme(legend.position = "none"),
                        small_block_pub, confusion_legend_pub,
                        nrow = 1, rel_widths = c(1.0, 1.42, 0.17))

liver_label_data <- copy(liver_xcell)[order(-shap_pct)][seq_len(min(9, .N))]

plot_E_scatter <- ggplot(liver_xcell, aes(top1pct, shap_pct)) +
  geom_smooth(method = "lm", se = FALSE, colour = "grey55",
              linetype = "dashed", linewidth = 0.35) +
  geom_point(aes(fill = dominant_ct), size = 1.0, shape = 21,
             colour = "grey25", stroke = 0.2) +
  ggrepel::geom_text_repel(data = liver_label_data, aes(label = LV),
                           size = MIN_GEOM_TEXT, max.overlaps = Inf,
                           min.segment.length = 0, segment.size = 0.18,
                           segment.color = "grey65", box.padding = 0.15) +
  annotate("text", x = -Inf, y = Inf, label = r_label_liver,
           hjust = -0.05, vjust = 1.2, size = MIN_GEOM_TEXT, parse = TRUE) +
  annotate("text", x = -Inf, y = Inf, label = p_label_liver,
           hjust = -0.05, vjust = 2.4, size = MIN_GEOM_TEXT, parse = TRUE) +
  scale_fill_manual(values = CT_COLORS_LIVER, labels = CT_LABELS_LIVER,
                    name = "Dominant\ncell type") +
  labs(x = "Mean hepatocyte xCell score in top-1% GTEx samples",
       y = "Individual SHAP importance (%)") +
  theme_nm() +
  guides(fill = guide_legend(ncol = 1, byrow = TRUE, title.position = "top",
                             title.hjust = 0.5,
                             override.aes = list(size = 1.8))) +
  theme(axis.text = element_text(size = MIN_FONT_PT),
        axis.title = element_text(size = 5.8),
        legend.position = "right", legend.direction = "vertical",
        legend.text = element_text(size = MIN_FONT_PT),
        legend.title = element_text(size = MIN_FONT_PT, hjust = 0.5),
        legend.key.width = unit(2.4, "mm"),
        legend.key.height = unit(2.0, "mm"),
        legend.spacing.x = unit(0.5, "mm"),
        plot.margin = margin(1, 0.3, 0.5, 1.5))

scatter_legend_pub <- get_legend(plot_E_scatter)
plot_F_scatter <- plot_grid(
  plot_E_scatter + theme(legend.position = "none"),
  scatter_legend_pub,
  nrow = 1, rel_widths = c(1, 0.29)
)

pathway_pub <- copy(pathway_sel)
pathway_pub[, pathway_display := pathway_clean]
pathway_pub[grepl("acyl-CoA Oxidase", pathway_display, ignore.case = TRUE),
            pathway_display := "FA beta-oxidation (acyl-CoA oxidase)"]
pathway_pub[grepl("acyl-CoA Dehydrog", pathway_display, ignore.case = TRUE),
            pathway_display := "FA beta-oxidation (acyl-CoA dehydrog.)"]
pathway_pub[, pathway_display := shorten_label(pathway_display, width = 43)]
pathway_pub[, row_key_pub := factor(paste0(pathway_display, "___", LV),
                                    levels = unique(paste0(pathway_display, "___", LV)))]

plot_E_pathways <- ggplot(pathway_pub, aes(AUC, row_key_pub)) +
  geom_segment(aes(x = 0.45, xend = AUC, yend = row_key_pub),
               color = "grey78", linewidth = 0.25, linetype = "dotted") +
  geom_point(aes(color = neg_log_fdr, size = AUC)) +
  scale_color_gradientn(colors = GREEN_PATHWAY, limits = c(0, 3),
                        name = "-log10(FDR)") +
  guides(color = guide_colorbar(title.position = "top",
                                 barheight = unit(10, "mm"),
                                 barwidth = unit(1.6, "mm"))) +
  scale_size_continuous(range = c(0.6, 1.8), guide = "none") +
  scale_x_continuous(limits = c(0.45, 1), breaks = seq(0.5, 1, 0.25)) +
  scale_y_discrete(labels = strip_lv_suffix) +
  facet_wrap(~LV, ncol = 1, scales = "free_y") +
  labs(x = "AUC", y = "GO:BP") +
  theme_nm() +
  theme(panel.grid.major.y = element_line(color = "grey88", linewidth = 0.18,
                                          linetype = "dotted"),
        strip.background = element_blank(),
        strip.text = element_text(size = 5.3, face = "bold", margin = margin(0.4, 0, 0.4, 0)),
        axis.text.x = element_text(size = MIN_FONT_PT),
        axis.text.y = element_text(size = MIN_FONT_PT),
        axis.title = element_text(size = 5.8),
        legend.position = "right", legend.text = element_text(size = MIN_FONT_PT),
        legend.title = element_text(size = 5),
        plot.margin = margin(0.5, 0.5, 0.8, 1.5))

plot_G_pub <- plot_grid(plot_F_scatter, NULL, plot_E_pathways, ncol = 1,
                        rel_heights = c(1.52, 0.10, 0.68))

# Additional Figure 2 panels assembled from canonical workflow exports.

FONT_FAMILY <- "Helvetica"

# E: ranked CLAMPfull gene loadings for the four displayed hard cell-type pairs.
loading_ready <- fread_gm(panel_input[["hard_pair_loading_ready"]])
loading_stats <- fread_gm(panel_input[["hard_pair_loading_stats"]])
loading_stats <- loading_stats[shown == TRUE]
stopifnot(unique(loading_stats$marker_source) == "Azimuth 2023",
          unique(loading_ready$marker_source) == "Azimuth 2023")
loading_ready <- loading_ready[pair_id %in% loading_stats$pair_id & loading > 0]

LOADING_OTHER <- "#D0D0D0"

compact_pair_title <- function(x) {
  if (grepl("^01_", x)) return(PAIR_DISPLAY_TITLES[["1"]])
  if (grepl("^02_", x)) return(PAIR_DISPLAY_TITLES[["2"]])
  if (grepl("^04_", x)) return(PAIR_DISPLAY_TITLES[["4"]])
  if (grepl("^06_", x)) return(PAIR_DISPLAY_TITLES[["6"]])
  x <- sub("^[0-9]+_", "", x)
  x <- gsub("_vs_", " vs. ", x)
  x <- gsub("Atrial vs. Ventricular", "Atrial vs. ventricular CM", x)
  x <- gsub("Ast vs. Mic", "Astrocyte vs. microglia", x)
  x <- gsub("CD4 vs. CD8", "CD4+ vs. CD8+ T cells", x)
  x <- gsub("B vs. T", "B vs. T cells", x)
  x
}

compact_loading_ct <- function(x) {
  x <- gsub("_", " ", x, fixed = TRUE)
  x <- gsub("VentricularCM", "Ventricular CM", x, fixed = TRUE)
  x <- gsub("AtrialCM", "Atrial CM", x, fixed = TRUE)
  x <- gsub("B cell", "B cell", x, fixed = TRUE)
  x
}

format_loading_p <- function(p) {
  if (is.na(p)) return("P n.a.")
  if (p < 1e-2) {
    exponent <- floor(log10(p))
    mantissa <- p / 10^exponent
    superscript <- chartr("-0123456789", "⁻⁰¹²³⁴⁵⁶⁷⁸⁹",
                         as.character(exponent))
    return(sprintf("P=%.1f×10%s", mantissa, superscript))
  }
  sprintf("P=%.2f", p)
}

make_loading_side <- function(pid, side) {
  y_max <- c(`01_B_vs_T`=1.5, `02_CD4_vs_CD8`=3,
             `04_Ast_vs_Mic`=1.5, `06_Atrial_vs_Ventricular`=1.5)[[pid]]
  d <- copy(loading_ready[pair_id == pid & panel_side == side])
  s <- loading_stats[pair_id == pid & panel_side == side]
  stopifnot(nrow(s) == 1L)
  first_set <- canonical_pair_set(s$set_A_label)
  second_set <- canonical_pair_set(s$set_B_label)
  d[, set_name := fifelse(grp == "A", first_set,
                   fifelse(grp == "B", second_set,
                     fifelse(grp == "shared", "Shared marker", "Other")))]
  d[, set_name := factor(set_name, levels = c("Other", PAIR_LEGEND_ORDER))]
  recovered_set <- if (side == "A") first_set else second_set
  n_recovered <- d[as.character(set_name) == recovered_set, .N]
  n_available <- if (side == "A") s$n_A_excl else s$n_B_excl
  label_d <- d[is_label == TRUE][order(rank)][seq_len(min(.N, 5L))]
  # One row per label; right-half points get a right-aligned label above them.
  label_d[, lab_y := y_max * (0.93 - 0.13 * (seq_len(.N) - 1))]
  label_d[, right := rank > max(d$rank) * 0.5]
  label_d[, `:=`(lab_x = fifelse(right, rank, max(d$rank) * 0.28),
                 lab_h = fifelse(right, 1, 0),
                 seg_y = fifelse(right, lab_y - y_max * 0.05, lab_y))]

  loading_plot <- ggplot(d, aes(rank, loading)) +
    ggrastr::rasterise(geom_point(data = d[set_name == "Other"], colour = LOADING_OTHER,
               size = 0.12, alpha = 0.65), dpi=300) +
    geom_point(data = d[set_name != "Other"], aes(colour = set_name),
               size = 0.48, alpha = 0.9) +
    # Fixed label column above the curve (ordered by rank, so leader lines do not cross);
    # ggrepel measured the text with a different font than the one drawn and overlapped.
    geom_segment(data = label_d, aes(x = rank, y = loading, xend = lab_x, yend = seg_y,
                                     colour = set_name),
                 linewidth = 0.14, alpha = 0.45, show.legend = FALSE) +
    geom_text(data = label_d, aes(x = lab_x, y = lab_y, label = gene, colour = set_name,
                                  hjust = lab_h), vjust = 0.5, size = MIN_GEOM_TEXT, fontface = "italic",
              family = FONT_FAMILY, show.legend = FALSE) +
    annotate("text", x = Inf, y = Inf,
             label = sprintf("%d/%d", n_recovered, n_available),
             colour = PAIR_SET_COLORS[[recovered_set]],
             hjust = 1.05, vjust = 1.25, size = MIN_GEOM_TEXT,
             fontface = "bold", family = FONT_FAMILY) +
    scale_colour_manual(
      values = PAIR_SET_COLORS,
      breaks = PAIR_LEGEND_ORDER,
      drop = FALSE, name = "Azimuth gene set"
    ) +
    scale_x_continuous(breaks=function(lims) {
      ticks <- pretty(lims, n=3)
      ticks[ticks >= 0 & ticks <= max(d$rank)]
    }, expand = expansion(mult = c(0.015, 0.08))) +
    scale_y_continuous(limits=c(0,y_max),breaks=seq(0,y_max,by=0.5),
                       expand=expansion(mult=c(0,.02))) +
    labs(
      x = NULL, y = NULL,
      title = NULL,
      subtitle = sprintf("AUC %s\n%s",
                         if (pid == "01_B_vs_T" && side == "A")
                           sprintf("%.3f/%.3f", s$auc_own, s$auc_other)
                         else sprintf("%.2f/%.2f", s$auc_own, s$auc_other),
                         format_loading_p(s$wilcox_p))
    ) +
    theme_nm() +
    theme(
      axis.text = element_text(size = MIN_FONT_PT),
      axis.title = element_blank(),
      plot.title = element_text(size = MIN_FONT_PT, face = "bold", hjust = 0.5,
                                margin = margin(b = 0.1)),
      plot.subtitle = element_text(size = MIN_FONT_PT, colour = "black", hjust = 0.5,
                                   lineheight = 0.88, margin = margin(b = 0.3)),
      legend.position = "none",
      plot.margin = margin(0.6, 0.7, 0.7, 0.7, "mm")
    )
  header_label <- sprintf("%s %s",compact_loading_ct(s$cell_type),s$LV)
  # Place the title in the same gtable columns as the AUC/P subtitle.
  header_text <- grid::textGrob(header_label, x=unit(.5,"npc"),y=unit(.5,"npc"),
    just="centre",gp=grid::gpar(fontsize=MIN_FONT_PT,
      fontfamily=FONT_FAMILY,fontface="bold"))
  header_dot <- grid::circleGrob(
    x=unit(.5,"npc")-grid::grobWidth(header_text)/2-unit(1.1,"mm"),
    y=unit(.5,"npc"),r=unit(.65,"mm"),
    gp=grid::gpar(fill=PAIR_SET_COLORS[[recovered_set]],col=NA))
  loading_grob <- ggplotGrob(loading_plot)
  subtitle_slot <- loading_grob$layout[loading_grob$layout$name == "subtitle", ]
  stopifnot(nrow(subtitle_slot) == 1L)
  loading_grob <- gtable::gtable_add_rows(loading_grob, unit(2.7,"mm"),
                                        pos=subtitle_slot$t-1L)
  gtable::gtable_add_grob(loading_grob, grid::grobTree(header_text,header_dot),
                        t=subtitle_slot$t, l=subtitle_slot$l, r=subtitle_slot$r,
                        clip="off", name="module_title")
}

loading_pair_ids <- intersect(
  c("01_B_vs_T", "02_CD4_vs_CD8", "04_Ast_vs_Mic",
    "06_Atrial_vs_Ventricular"),
  unique(loading_stats$pair_id)
)
stopifnot(length(loading_pair_ids) == 4L)
loading_pair_blocks <- lapply(loading_pair_ids, function(pid)
  plot_grid(make_loading_side(pid, "A"), make_loading_side(pid, "B"),
            nrow = 1, align = "h", axis = "tb"))

loading_grid <- plot_grid(plotlist = loading_pair_blocks, ncol = 2)
loading_grid_labeled <- ggdraw() +
  draw_plot(loading_grid, x = 0.04, y = 0.075, width = 0.96, height = 0.925) +
  draw_label("Genes with Z loading > 0, ranked by loading", x = 0.51, y = 0.060,
             hjust = 0.5, vjust = 0, size = 5.2, fontfamily = FONT_FAMILY) +
  draw_label("CLAMPfull gene loading", x = 0.012, y = 0.54, angle = 90,
             hjust = 0.5, vjust = 0.5, size = 5.2, fontfamily = FONT_FAMILY)
plot_E_loading <- loading_grid_labeled

# G: multitissue xCell/CellMarker recovery from notebook 06's panel-ready export.
multi <- fread_gm(panel_input[["multitissue_panel_ready"]])
multi[, short := factor(short, levels = unique(short))]
multi[, cell_type := factor(cell_type, levels = unique(cell_type))]
G2_SHORT_LABELS <- c(Epithelial="Epithel.", Hepatocytes="Hepat.",
 Neurons="Neuron", Myocytes="Myocyte", Adipocytes="Adipoc.",
 Neutrophils="Neutroph.", Keratinocytes="Keratin.")
multi_lv <- unique(multi[, .(tissue, short, LV, shap_pct, rank, ora_term, ora_fold,
                              ora_genes, ora_neg_log10_fdr)])
setorder(multi_lv, short, rank)
multi_lv[, term_clean := fifelse(is.na(ora_term) | !nzchar(ora_term), "n.s.", ora_term)]
multi_lv[, term_clean := gsub(" cell$", "", term_clean, ignore.case=TRUE)]
multi_lv[, term_clean := gsub("Liver bud hepatic.*", "Liver bud", term_clean)]
multi_lv[, term_clean := gsub("DCLK1.*", "DCLK1+ progenitor", term_clean)]
multi_lv[, key := paste(tissue, LV)]
multi_row_labels <- rev(sprintf("%s (%.1f%%) · %s", multi_lv$LV, multi_lv$shap_pct,
                                multi_lv$term_clean))
multi_lv[, row_id := factor(key, levels = rev(key), labels = multi_row_labels)]
multi[, row_id := factor(paste(tissue, LV), levels = rev(multi_lv$key),
                         labels = multi_row_labels)]

# Thin dashed rule below every tissue group except the last (shared by both halves of g).
multi_seps <- unique(multi_lv[, .(short)])[-.N]
band_layer <- function() geom_hline(data = multi_seps, aes(yintercept = 0.5), colour = "black",
                                    linewidth = 0.2, linetype = "22", inherit.aes = FALSE)
tissue_label <- function(x) {
  x <- gsub(" ", "\n", x, fixed = TRUE)
  paste0(substr(x, 1, 1), tolower(substring(x, 2)))
}
multi_ora <- ggplot(multi_lv, aes(ora_fold, row_id)) +
  band_layer() +
  geom_segment(aes(x = 0, xend = ora_fold, yend = row_id), colour = "grey70",
               linewidth = 0.3, na.rm = TRUE) +
  geom_point(aes(colour = ora_neg_log10_fdr, size = ora_genes), na.rm = TRUE) +
  facet_grid(short ~ ., scales = "free_y", space = "free_y", switch = "y",
             labeller = labeller(short = tissue_label)) +
  scale_colour_gradientn(colours = GREEN_PATHWAY, breaks = c(5, 15, 25),
                         name = "-log10(FDR)") +
  scale_size_continuous(range = c(0.55, 2.0), name = "Genes", breaks = c(5, 15, 25)) +
  guides(
    colour = guide_colourbar(order = 1, title.position = "top", title.hjust = 0.5,
                             barwidth = unit(16, "mm"),
                             barheight = unit(1.2, "mm")),
    size = guide_legend(order = 2, title.position = "top", nrow = 1,
                        override.aes = list(size = c(0.8, 1.2, 1.7)))
  ) +
  scale_x_continuous(breaks = c(0, 15, 30), expand = expansion(mult = c(0, 0.10))) +
  labs(x = "Fold enrichment", y = NULL) +
  theme_nm() +
  theme(strip.placement = "outside", strip.background = element_blank(),
        strip.text.y.left = element_text(angle = 0, face = "bold", size = MIN_FONT_PT,
                                         hjust = 1, margin = margin(r = 1.2, unit = "mm")),
        axis.text.y = element_text(size = MIN_FONT_PT, lineheight = 0.88),
        axis.text.x = element_text(size = MIN_FONT_PT), axis.title.x = element_text(size = MIN_FONT_PT),
        legend.position = "bottom", legend.box = "horizontal",
        legend.box.just = "center", legend.spacing.x = unit(0.6, "mm"),
        legend.spacing.y = unit(0, "mm"),
        legend.text = element_text(size = MIN_FONT_PT), legend.title = element_text(size = MIN_FONT_PT),
        legend.key.width = unit(1.8, "mm"), legend.key.height = unit(1.35, "mm"),
        legend.margin = margin(0, 0, 0, 0),
        panel.spacing.y = unit(0.2, "mm"),
        plot.margin = margin(0.5, 0.5, 0.5, 0.5))

multi_fill_vals <- c(0, .20, .375, .4583, .4584, .5416, .5417, .625, .80, 1)
multi_fill_cols <- c("#67001f", "#b2182b", "#d6604d", "#f4a582", "white", "white",
                     "#a1d9b6", "#57b37e", "#007a33", "#00431e")
multi_xcell <- ggplot(multi, aes(cell_type, row_id)) +
  band_layer() +
  # Physical-size square outlines enclose the circles without spanning the grid cell.
  geom_point(data = multi[is_dominant == TRUE], shape = 0, colour = "grey20",
             size = 2.4, stroke = 0.25) +
  geom_point(aes(size = top1pct, fill = l2fc_shap), shape = 21, stroke = 0.12,
             colour = "grey40") +
  facet_grid(short ~ ., scales = "free_y", space = "free_y",
             labeller = labeller(short = function(x)
               gsub(" ", "\n", x, fixed = TRUE))) +
  scale_x_discrete(labels=G2_SHORT_LABELS) +
  scale_size_binned(name = "xCell score", breaks = c(.1, .25, .5, .9),
                    limits = c(0, max(multi$top1pct)), range = c(.35, 2.0)) +
  scale_fill_gradientn(colours = multi_fill_cols, values = multi_fill_vals,
                       limits = c(-6, 6), oob = scales::squish,
                       breaks = c(-6, -3, 0, 3, 6),
                       labels = c("≤-6", "-3", "0", "3", "≥6"),
                       name = "log2 FC SHAP") +
  guides(
    size = guide_legend(order = 1, title.position = "top", nrow = 1,
                        override.aes = list(size = c(0.7, 1.0, 1.4, 1.8))),
    fill = guide_colourbar(order = 2, title.position = "top", title.hjust = 0.5,
                           barwidth = unit(16, "mm"),
                           barheight = unit(1.2, "mm"))
  ) +
  labs(x = NULL, y = NULL) +
  theme_nm() +
  theme(axis.text.x = element_text(angle = 40, hjust = 1, size = MIN_FONT_PT),
        axis.text.y = element_blank(), axis.ticks.y = element_blank(),
        strip.background = element_blank(), strip.text.y = element_blank(),
        legend.position = "bottom", legend.box = "horizontal",
        legend.box.just = "center", legend.spacing.x = unit(0.6, "mm"),
        legend.spacing.y = unit(0, "mm"),
        legend.text = element_text(size = MIN_FONT_PT), legend.title = element_text(size = MIN_FONT_PT),
        legend.key.width = unit(1.8, "mm"), legend.key.height = unit(1.35, "mm"),
        legend.margin = margin(0, 0, 0, 0),
        panel.spacing.y = unit(0.2, "mm"),
        plot.margin = margin(0.5, 0.5, 0.5, 0.5))




# G/H: donor-bulk held-out recovery and projection back to PBMC Perez cells.
donor_predictions <- fread_gm(panel_input[["donor_bulk_predictions"]])
donor_statistics <- fread_gm(panel_input[["donor_bulk_statistics"]])
perez_annotation <- fread_gm(panel_input[["donor_bulk_perez_annotation"]])
perez_activity <- fread_gm(panel_input[["donor_bulk_perez_activity"]])
perez_summary <- fread_gm(panel_input[["donor_bulk_perez_summary"]])

donor_predictions[, panel_label := DATASET_LABELS[dataset]]
donor_plot_statistics <- copy(donor_statistics)
stopifnot(all(is.finite(donor_plot_statistics$mean_pearson_r)), all(is.finite(donor_plot_statistics$mean_mae)))
donor_plot_statistics[, panel_label := DATASET_LABELS[dataset]]
donor_predictions[, panel_label := factor(panel_label, levels=panel_levels)]
donor_plot_statistics[, panel_label := factor(panel_label, levels=panel_levels)]
donor_plot_statistics[, r2_label := sprintf('r = %.2f, MAE = %.3f', mean_pearson_r, mean_mae)]
donor_plot_statistics[, facet_label := factor(sprintf("%s\nr=%.2f  MAE=%.3f",
  c("1k1k","Perez","Heart","Xiong","Mathys","Lung")[match(as.character(panel_label), c("PBMC 1k1k","PBMC Perez","Heart Datar","Brain Xiong","Brain Mathys","Lung Sikkema"))], mean_pearson_r, mean_mae))]
donor_facet_lookup <- setNames(as.character(donor_plot_statistics$facet_label),
  as.character(donor_plot_statistics$panel_label))
donor_predictions[, facet_label := factor(donor_facet_lookup[as.character(panel_label)],
  levels=as.character(donor_plot_statistics$facet_label))]
plot_G_scatter <- ggplot(donor_predictions, aes(observed, predicted)) +
  geom_abline(slope=1, intercept=0, color="grey25",
              linetype="dashed", linewidth=0.25) +
  ggrastr::rasterise(geom_point(aes(color=panel_label), alpha=0.28, size=0.16,
             show.legend=FALSE), dpi=300) +
  facet_wrap(~facet_label, ncol=3) +
  coord_cartesian(xlim=c(0,1), ylim=c(0,1), expand=FALSE) +
  scale_x_continuous(breaks=c(0,.5,1),
    labels=c("0","0.5","1")) +
  scale_y_continuous(breaks=c(0,.5,1),
    labels=c("0","0.5","1")) +
  scale_color_manual(values=DATASET_COLORS) +
  labs(x="True donor cell proportion", y="Predicted donor cell proportion") +
  theme_nm() +
  theme(aspect.ratio=.48, axis.text=element_text(size=MIN_FONT_PT),
        axis.title=element_text(size=6),
        panel.spacing.x=unit(1.0, "mm"),
        panel.spacing.y=unit(1.7, "mm"),
        strip.text=element_text(size=5, face="bold", lineheight=.85, margin=margin(b=0.3)),
        plot.margin=margin(2.0, 0.3, 0.5, 0.7))

PEREZ_COLORS <- c(B_cell = "#4477AA", Myeloid = "#EE6677",
                  NK = "#228833", T_cell = "#CCBB44")
PEREZ_LABELS <- c(B_cell = "B cell", Myeloid = "Myeloid",
                  NK = "NK", T_cell = "T cell")
perez_annotation[, mapped_cell_type := factor(mapped_cell_type, levels = names(PEREZ_COLORS))]
perez_centroids <- perez_annotation[, .(umap1 = median(umap1), umap2 = median(umap2)),
                                     by = mapped_cell_type]
perez_xr <- range(perez_annotation$umap1, finite = TRUE)
perez_yr <- range(perez_annotation$umap2, finite = TRUE)
perez_span <- max(diff(perez_xr), diff(perez_yr)) * 1.08
perez_xlim <- mean(perez_xr) + c(-0.5, 0.5) * perez_span
perez_ylim <- mean(perez_yr) + c(-0.5, 0.5) * perez_span
plot_H_annotation <- ggplot(perez_annotation, aes(umap1, umap2, colour = mapped_cell_type)) +
  ggrastr::rasterise(geom_point(size = 0.18, alpha = 0.80), dpi = 300) +
  scale_colour_manual(values = PEREZ_COLORS, labels = PEREZ_LABELS, name = NULL) +
  coord_equal(xlim = perez_xlim, ylim = perez_ylim, expand = FALSE) +
  labs(title = "Cell-type annotation", x = "UMAP 1", y = "UMAP 2") +
  theme_classic(base_family = FONT_FAMILY, base_size = MIN_FONT_PT) +
  theme(legend.position = "right", legend.text = element_text(size = MIN_FONT_PT),
        legend.key.size = unit(1.5, "mm"), legend.spacing.y = unit(0, "mm"),
        axis.text = element_text(size = MIN_FONT_PT),
        axis.title = element_text(size = MIN_FONT_PT),
        plot.title = element_text(size = MIN_FONT_PT, face = "bold", hjust = .5),
        plot.margin = margin(.5,.5,.5,.5,"mm"))
PEREZ_SHORT_H <- c(B_cell="B", Myeloid="Mye", NK="NK", T_cell="T")
perez_types_H <- names(PEREZ_COLORS)
plot_H_features <- lapply(seq_along(perez_types_H), function(i) {
  ct <- perez_types_H[[i]]
  dat <- perez_activity[cell_type == ct]
  stopifnot(nrow(dat) > 0L, uniqueN(dat$LV) == 1L)
  ggplot(dat, aes(umap1, umap2, colour = activity)) +
    ggrastr::rasterise(geom_point(size = 0.18), dpi = 300) +
    scale_colour_gradientn(colours = c("#D8E2EF", "white", PEREZ_COLORS[[ct]])) +
    scale_x_continuous(breaks = c(-5, 10)) +
    scale_y_continuous(breaks = c(-5, 10)) +
    coord_equal(xlim = perez_xlim, ylim = perez_ylim, expand = FALSE) +
    labs(title = sprintf("%s · %s", PEREZ_SHORT_H[[ct]], unique(dat$LV)),
         x = "UMAP 1",
         y = if (i == 1) "UMAP 2" else NULL) +
    theme_classic(base_family = FONT_FAMILY, base_size = MIN_FONT_PT) +
    theme(legend.position = "none",
          axis.text.x = element_text(size = MIN_FONT_PT),
          axis.text.y = if (i == 1) element_text(size = MIN_FONT_PT) else element_blank(),
          axis.ticks.y = if (i == 1) element_line() else element_blank(),
          axis.title = element_text(size = MIN_FONT_PT),
          axis.ticks.length = unit(.45, "mm"),
          plot.title = element_text(size = MIN_FONT_PT, face = "bold", hjust = .5),
          plot.margin = margin(.5,.5,.5,.5,"mm"))
})
perez_annotation_legend_H <- get_legend(
  plot_H_annotation + guides(colour=guide_legend(
    nrow=1,byrow=TRUE,override.aes=list(size=2))) +
    theme(legend.position="bottom",legend.direction="horizontal",
          legend.text=element_text(size=MIN_FONT_PT),
          legend.key.width=unit(1.8,"mm"),
          legend.spacing.x=unit(0.2,"mm")))
perez_features_H <- ggdraw() +
  draw_plot(plot_grid(plotlist=plot_H_features,nrow=1,align="hv"),
            x=0,y=0,width=1,height=1) +
  draw_grob(perez_annotation_legend_H,x=.12,y=.12,width=.76,height=.11)
perez_maps_core_H <- plot_grid(
  plot_H_annotation + theme(legend.position="none"), perez_features_H,
  nrow=1,rel_widths=c(0.36,0.64))
perez_maps_H <- perez_maps_core_H
perez_auc_H <- perez_summary$median_auroc[[1]]
perez_r_H <- perez_summary$median_bulk_r[[1]]
perez_header_H <- ggdraw() +
  draw_label(sprintf("PBMC: Perez     AUROC = %.2f     r = %.2f",
                     perez_auc_H, perez_r_H),
             x = 0.5, y = 0.5, hjust = 0.5, size = MIN_FONT_PT,
             fontface = "bold", fontfamily = FONT_FAMILY)
perez_block_H <- plot_grid(perez_header_H, perez_maps_H, ncol = 1,
                           rel_heights = c(0.16, 0.84))

cross_grid <- fread_gm(panel_input[["hard_cross_grid"]])
cross_grid <- cross_grid[dataset != "Lung_Sikkema2023" &
  !(dataset == "PBMC_1k1k" & lv_cell_type == "B_cell")]
stopifnot(nrow(cross_grid)==80L, uniqueN(cross_grid$module_id)==10L,
          uniqueN(cross_grid$marker_id)==8L)
cohort_names_C <- c(Heart_Datar2026="Heart", PBMC_1k1k="1k1k",
  Lung_Sikkema2023="Lung", Brain_Mathys2023="Mathys",
  Brain_Xiong2023="Xiong", PBMC_Perez2022="Perez")
cohort_colors_C <- c(Heart_Datar2026=DATASET_COLORS[["Heart Datar"]],
  PBMC_1k1k=DATASET_COLORS[["PBMC 1k1k"]],
  Lung_Sikkema2023=DATASET_COLORS[["Lung Sikkema"]],
  Brain_Mathys2023=DATASET_COLORS[["Brain Mathys"]],
  Brain_Xiong2023=DATASET_COLORS[["Brain Xiong"]],
  PBMC_Perez2022=DATASET_COLORS[["PBMC Perez"]])
marker_labels <- c("B cell", "T cell", "CD4+ T", "CD8+ T",
                   "Astrocyte", "Microglia", "Atrial CM", "Ventricular CM")
module_cell_label_C <- c(AtrialCM="Atrial CM", VentricularCM="Ventricular CM",
  B_cell="B cell", CD4_T="CD4+ T", CD8_T="CD8+ T",
  Lymphoid="Lymphoid", Myeloid="Myeloid", Ast="Astrocyte",
  Mic="Microglia", T_cell="T cell")
cohort_order_C <- c("PBMC_Perez2022", "PBMC_1k1k",
  "Brain_Mathys2023", "Brain_Xiong2023", "Heart_Datar2026",
  "Lung_Sikkema2023")
module_meta_C <- unique(cross_grid[,.(module_rank,dataset,LV,lv_cell_type)])
module_meta_C[, cohort_rank := match(dataset, cohort_order_C)]
setorder(module_meta_C,cohort_rank,module_rank)
stopifnot(nrow(module_meta_C)==10L,
  all(module_meta_C$lv_cell_type %in% names(module_cell_label_C)))
module_meta_C[, x_new := seq_len(.N)]
module_axis_C <- sprintf("%s/%s", module_meta_C$LV,
  module_cell_label_C[module_meta_C$lv_cell_type])
cross_grid[, x := module_meta_C$x_new[match(module_rank,module_meta_C$module_rank)]]
cross_grid[, `:=`(y=9-marker_rank,
  ora_fill=ifelse(FDR<.01,pmin(neg_log10_fdr,20),0))]
format_fdr_C <- function(q) {
  if (q >= 1) return("1")
  e <- floor(log10(q))
  m <- q / 10^e
  if (round(m) >= 10) { m <- 1; e <- e + 1 }
  superscript <- chartr("-0123456789", "⁻⁰¹²³⁴⁵⁶⁷⁸⁹", as.character(e))
  sprintf("%.0f×10%s", m, superscript)
}
cross_grid[, ora_label := vapply(FDR, format_fdr_C, character(1))]
header_C <- module_meta_C[,.(xmin=min(x_new)-.5,
  xmax=max(x_new)+.5),by=dataset]
header_C[, `:=`(xmid=(xmin+xmax)/2,ymin=8.55,ymax=9.25,
  label=cohort_names_C[dataset],color=cohort_colors_C[dataset],
  text_color=ifelse(dataset %in% c("PBMC_Perez2022", "PBMC_1k1k",
    "Heart_Datar2026", "Brain_Xiong2023"), "white", "black"))]
setorder(header_C,xmin)
plot_E_pub <- ggplot(cross_grid,aes(x,y,fill=ora_fill)) +
 geom_tile(width=.98,height=.98,colour="white",linewidth=.16) +
 geom_tile(data=cross_grid[matched==TRUE],width=.98,height=.98,
           fill=NA,colour="black",linewidth=.35) +
 geom_segment(data=header_C[-nrow(header_C)],
  aes(x=xmax,y=.5,xend=xmax,yend=8.5),inherit.aes=FALSE,
  colour="grey30",linewidth=.22) +
 geom_rect(data=header_C,
  aes(xmin=xmin,xmax=xmax,ymin=ymin,ymax=ymax),inherit.aes=FALSE,
  fill=header_C$color,colour="white",linewidth=.2) +
 geom_text(data=header_C,aes(x=xmid,y=8.9,label=label),inherit.aes=FALSE,
  size=MIN_GEOM_TEXT,fontface="bold",family=FONT_FAMILY,
  colour=header_C$text_color) +
 geom_text(data=cross_grid[FDR < .01 & ora_fill < 10], aes(label=ora_label), lineheight=.55,
  size=MIN_GEOM_TEXT, colour="black", family=FONT_FAMILY) +
 geom_text(data=cross_grid[FDR < .01 & ora_fill >= 10], aes(label=ora_label), lineheight=.55,
  size=MIN_GEOM_TEXT, colour="white", family=FONT_FAMILY) +
 scale_fill_gradientn(
  colours=c("#F3F4F6","#F3F4F6","#E3F2E7","#94CAA8","#007A33"),
  values=c(0,2/20,2.00001/20,10/20,1),limits=c(0,20),
  breaks=c(0,5,10,20),labels=c("0","5","10","≥20"),
  name="-log10(FDR)") +
 scale_x_continuous(breaks=seq_len(nrow(module_meta_C)),labels=module_axis_C,
  limits=c(.5,nrow(module_meta_C)+.5),expand=c(0,0)) +
 scale_y_continuous(breaks=8:1,labels=marker_labels,
  limits=c(.5,9.3),expand=c(0,0)) +
 labs(x="Gene module", y="Marker gene set") + theme_nm() +
 theme(axis.title=element_text(size=MIN_FONT_PT)) +
 theme(axis.line=element_blank(),axis.ticks=element_blank(),
       axis.text.y=element_text(size=MIN_FONT_PT),
       axis.text.x=element_text(angle=45,hjust=1,vjust=1,size=MIN_FONT_PT,
                                lineheight=.9),
       legend.position="bottom",legend.title=element_text(size=MIN_FONT_PT),
       legend.text=element_text(size=MIN_FONT_PT),
       plot.margin=margin(1,.3,.5,.7,"mm")) +
 guides(fill=guide_colourbar(title.position="top",barwidth=unit(15,"mm"),
                             barheight=unit(1.1,"mm")))
row_AB_pub <- plot_grid(add_tag(plot_A_pub, "a"),
                        add_tag(plot_C_pub, "b"),
                        nrow = 1, rel_widths = c(0.47, 0.53))
module_compact_panels <- lapply(module_pub_panels, function(p)
  p + theme(legend.position = "none",
            axis.text.x = element_text(angle = 90, hjust = 1, vjust = 0.5, size = MIN_FONT_PT),
            axis.text.y = element_text(size = MIN_FONT_PT),
            plot.title = element_text(size = MIN_FONT_PT)))
plot_E_marker_compact <- plot_grid(
  plot_grid(plotlist = module_compact_panels, ncol = 2),
  module_legend, ncol = 1, rel_heights = c(1, 0.075)
)
brain_core_G <- brain_pub +
  coord_fixed(ratio = 1.28) +
  labs(x = NULL, y = NULL) +
  theme(legend.position = "none", plot.margin = margin(0.2, 0.2, 0.2, 0.2))
small_grid_G <- wrap_plots(lapply(small_pub, function(p)
  p + coord_fixed(ratio = 1.22) +
    theme(legend.position = "none", axis.text = element_text(size = MIN_FONT_PT),
            plot.margin = margin(0.1, 0.1, 0.1, 0.1))), ncol = 2)
brain_legend_bottom_G <- get_legend(
  brain_pub +
    guides(fill = guide_colourbar(title = "True fraction",
                                  title.position = "top", title.hjust = 0.5,
                                  direction = "horizontal",
                                  barwidth = unit(15, "mm"),
                                  barheight = unit(1.2, "mm"))) +
    theme(legend.position = "bottom", legend.direction = "horizontal",
          legend.title = element_text(size = MIN_FONT_PT, hjust = 0.5),
          legend.text = element_text(size = MIN_FONT_PT),
          legend.margin = margin(0, 0, 0, 0))
)
G_matrix_body <- plot_grid(brain_core_G, small_grid_G, nrow = 1,
                           rel_widths = c(1.10, 0.90))
plot_G_confusion <- ggdraw() +
  draw_plot(G_matrix_body, x = 0.025, y = 0.10, width = 0.975, height = 0.90) +
  draw_label("Predicted region", x = 0.52, y = 0.125,
             size = MIN_FONT_PT, fontfamily = FONT_FAMILY) +
  draw_grob(brain_legend_bottom_G, x = 0.37, y = 0.005,
            width = 0.30, height = 0.085) +
  draw_label("True region", x = 0.012, y = 0.56, angle = 90,
             size = MIN_FONT_PT, fontfamily = FONT_FAMILY)
# Legends stacked in a column to the right of g, so the panel body is narrower.
legend_col_theme <- theme(legend.position = "right", legend.box = "vertical", legend.direction = "horizontal",
                          legend.box.just = "left", legend.justification = "left",
                          legend.spacing.y = unit(1.6, "mm"))
legend_ora_H <- get_plot_component(multi_ora + legend_col_theme, "guide-box-right")
legend_xcell_H <- get_plot_component(multi_xcell + legend_col_theme, "guide-box-right")
legend_outline_H <- ggdraw() +
  draw_grob(grid::rectGrob(x = unit(1.4, "mm"), width = unit(2.4, "mm"), height = unit(2.4, "mm"),
                           gp = grid::gpar(fill = NA, col = "grey20", lwd = .7))) +
  draw_grob(grid::circleGrob(x = unit(1.4, "mm"), r = unit(.75, "mm"),
                             gp = grid::gpar(fill = "white", col = "grey40", lwd = .45))) +
  draw_label("Highest xCell score", x = .12, hjust = 0, size = MIN_FONT_PT,
             fontfamily = FONT_FAMILY)
legend_H <- plot_grid(NULL, legend_ora_H, legend_xcell_H, legend_outline_H, NULL, ncol = 1,
                      rel_heights = c(.18, .85, 1, .16, .18), align = "v", axis = "l")
multi_body_H <- plot_grid(multi_ora+theme(legend.position="none"),
  multi_xcell+theme(legend.position="none"),nrow=1,
  rel_widths=c(1.45,1),align="h",axis="tb")
plot_H_xcell <- plot_grid(multi_body_H, NULL, legend_H, nrow = 1,
  rel_widths = c(.78, .03, .19))

row_CD_pub <- plot_grid(
  add_tag(plot_E_pub,"c"),
  plot_grid(NULL,add_tag(plot_E_loading,"d"),nrow=1,rel_widths=c(.018,.982)),
  nrow=1,rel_widths=c(.45,.55))
row_EF_pub <- plot_grid(add_tag(plot_G_scatter,"e"),add_tag(perez_block_H,"f"),
                        nrow=1,rel_widths=c(.48,.52))
fig2 <- plot_grid(row_AB_pub,row_CD_pub,row_EF_pub,NULL,
                  add_tag(ggdraw()+draw_plot(plot_H_xcell,x=.04,y=.01,width=.95,height=.98),"g"),
                  ncol=1,rel_heights=c(.2275,.2725,.2275,.018,.2545))

raw_pdf <- tempfile(fileext = ".pdf")
exact_pdf <- tempfile(fileext = ".pdf")
ggsave(raw_pdf, fig2,
       width = FIG_W, height = FIG_H, units = "mm",
       device = cairo_pdf, bg = "white", limitsize = FALSE)
pdf_w_pt <- FIG_W / 25.4 * 72
pdf_h_pt <- FIG_H / 25.4 * 72
gs_status <- system2("gs", c(
  "-q", "-dNOPAUSE", "-dBATCH", "-sDEVICE=pdfwrite",
  sprintf("-dDEVICEWIDTHPOINTS=%.6f", pdf_w_pt),
  sprintf("-dDEVICEHEIGHTPOINTS=%.6f", pdf_h_pt),
  "-dFIXEDMEDIA", "-dPDFFitPage",
  paste0("-sOutputFile=", exact_pdf), raw_pdf
))
stopifnot(gs_status == 0L, file.copy(exact_pdf, panel_output[["pdf"]],
                                      overwrite = TRUE))
unlink(c(raw_pdf, exact_pdf))
source(here::here("scripts/panels/svg_export.R"))
ggsave(panel_output[["svg"]], fig2,
       width = FIG_W, height = FIG_H, units = "mm",
       device = svg_device, bg = "white", limitsize = FALSE)
add_svg_font_stack(panel_output[["svg"]])
ggsave(panel_output[["png"]], fig2,
       width = FIG_W, height = FIG_H, units = "mm", dpi = 600,
       device = ragg::agg_png, bg = "white", limitsize = FALSE)

cat(sprintf("fig2: %.0f x %.0f mm, one-page Nature Methods main figure\n", FIG_W, FIG_H))
cat("output directory:", dirname(panel_output[["pdf"]]), "\n")
options(repr.plot.width = FIG_W / 25.4, repr.plot.height = FIG_H / 25.4)
fig2
```
